###   Exploration initiale des données foncières (DVF)

Avant d’entrer dans les analyses ciblées pour Sara, nous explorons les données foncières issues de la base DVF enrichie.  
Cette section permet d’obtenir une **vue d’ensemble du marché immobilier français**, à travers le calcul du **prix moyen au m²**, et d’identifier les **zones les plus chères ou les plus accessibles**.

---


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import ipywidgets as widgets
import plotly.graph_objects as go
import numpy as np

# Chargement du fichier enrichi
chunks = pd.read_csv("DataAll/Dataset.csv", chunksize=1000000, low_memory=False, sep=",")

chunk_list = []
for chunk in chunks:
    print("Processing a new chunk...")
    # Suppression des colonnes inutiles
    chunk = chunk.dropna(subset=["0"])
    chunk = chunk.drop(columns=["0", "code_postal_y"], errors="ignore")

    # Suppression des valeurs manquantes essentielles
    chunk = chunk.dropna(subset=["valeur_fonciere", "surface_reelle_bati"])

    # Calcul du prix au m²
    chunk["prix_m2"] = chunk["valeur_fonciere"] / chunk["surface_reelle_bati"]

    # Suppression des valeurs aberrantes
    chunk = chunk.replace([float('inf'), -float('inf')], None)
    chunk = chunk.dropna(subset=["prix_m2"])
    chunk = chunk[chunk["valeur_fonciere"] >= 50000 & (chunk["valeur_fonciere"] < 150000)]
    chunk = chunk[chunk["type_local"].isin(["Maison", "Appartement"])]
    chunk_list.append(chunk)

# Concatenate all chunks into a single DataFrame
dvf = pd.concat(chunk_list, ignore_index=True)

print("Dimensions:", dvf.shape)
```

### Prix moyen au m² par département

```python
dvf = dvf[(dvf["valeur_fonciere"] > 50000) & (dvf["valeur_fonciere"] < 150000)]
dvf = dvf[(dvf["surface_reelle_bati"] > 9) & (dvf["surface_reelle_bati"] < 1000)]
dvf = dvf[dvf["prix_m2"]>500 & (dvf["prix_m2"]<15000)]
```

```python
dvf = dvf[dvf["type_local"].isin(["Maison", "Appartement"])]
```

```python
fig = px.histogram(
    dvf, 
    x="prix_m2",
    nbins=40,
    animation_frame="Département",  #  crée un slider de départements
    color_discrete_sequence=["mediumturquoise"],
    title="Évolution de la distribution du prix au m² selon le département"
)

fig.update_layout(
    xaxis_title="Prix au m² (€)",
    yaxis_title="Nombre de biens vendus",
    showlegend=False
)

prix_dep = (
    dvf.groupby("Département")
       .agg(Prix_m2_median=("prix_m2", "median"),
            Prix_m2_mean=("prix_m2", "mean"),
            nb_ventes=("prix_m2", "size"))
       .sort_values("Prix_m2_median", ascending=False)
       .reset_index()
)

fig.show()
```

```python
prix_type = (
    dvf.groupby("type_local")
       .agg(Prix_m2_mean=("prix_m2", "mean"),
            nb_ventes=("prix_m2", "size"))
       .sort_values("Prix_m2_mean", ascending=False)
)
#  PIE CHART : même variables, seulement l'affichage change

labels = prix_type.index                     # type de bien (Maison, Appartement…)
sizes = prix_type["nb_ventes"]               # nombre de ventes = comme ton bar chart

explode = [0.03] * len(labels)               # légère séparation des parts

plt.figure(figsize=(7, 5))
plt.pie(
    sizes,
    labels=labels,
    autopct="%1.1f%%",       # pourcentage sur le graphique
    explode=explode,
    shadow=True,
    startangle=140
)

plt.title("Répartition de la variable type_local")
plt.axis("equal")            # cercle parfait
plt.show()
```

#### Interprétation

On observe que :
- Les **appartements** présentent le **prix au m² le plus élevé** (souvent situés en zone urbaine dense).
- Les **maisons** sont en moyenne **moins chères au m²**, mais leur surface totale est plus grande.
- Les **locaux commerciaux ou industriels** ont des prix plus variables selon la localisation.

➡️ Pour Sara, qui recherche un **investissement locatif clé en main**, les **appartements** (notamment T1 et T2) offrent un bon équilibre entre **valeur patrimoniale** et **facilité de location**.

## Analyse croisée : Prix moyen au m² vs Rendement locatif

Après avoir identifié les zones les plus chères, nous étudions maintenant la **relation entre le prix moyen au m²** et le **rendement locatif brut**.  
Cette approche permet de déterminer si le coût d’achat influence directement la rentabilité potentielle.

---



```python


# Suppression des valeurs aberrantes ou incomplètes
df_rend = dvf.dropna(subset=["valeur_fonciere", "surface_reelle_bati", "Loyerm2Mean", "Département"]).copy()
df_rend = df_rend[(df_rend["valeur_fonciere"] > 10000) & (df_rend["surface_reelle_bati"] > 10)]
dvf["rendement_brut"] = (df_rend["Loyerm2Mean"] * df_rend["surface_reelle_bati"] * 12) / df_rend["valeur_fonciere"]
# Calcul du rendement locatif brut annuel (%)
df_rend["rendement_brut"] = (
    (df_rend["Loyerm2Mean"] * df_rend["surface_reelle_bati"] * 12) / df_rend["valeur_fonciere"]
) * 100

# Suppression des valeurs aberrantes
df_rend = df_rend[(df_rend["rendement_brut"] >= 1) & (df_rend["rendement_brut"] <= 15)]

print(f"Nombre de transactions conservées : {len(df_rend)}")
df_rend[["Département", "Loyerm2Mean", "valeur_fonciere", "surface_reelle_bati", "rendement_brut"]].head()

# Calcul du rendement moyen et du prix médian au m² par département
rend_dep = (
    df_rend.groupby("Département")
    .agg(
        rendement_moyen=("rendement_brut", "mean"),
        prix_m2_median=("prix_m2", "median"),
        nb_ventes=("prix_m2", "size")
    )
    .sort_values("rendement_moyen", ascending=False)
    .reset_index()
)



```

###  Visualisation – Top 15 départements les plus rentables

Ce graphique présente les **15 départements les plus rentables** selon le rendement locatif brut moyen.
Les couleurs indiquent le **prix médian au m²**, afin de mettre en évidence le lien entre accessibilité et rentabilité.


```python
fig = px.scatter(
    rend_dep,
    x="prix_m2_median",
    y="rendement_moyen",
    size="nb_ventes",
    color="rendement_moyen",
    color_continuous_scale="Viridis",
    hover_name="Département",
    title="📈 Relation entre le prix au m² et le rendement locatif moyen par département",
    labels={
        "prix_m2_median": "Prix médian au m² (€)",
        "rendement_moyen": "Rendement brut moyen (%)"
    }
)

fig.update_layout(template="plotly_white")
fig.show()

```

```python
rend_dep = rend_dep[(rend_dep["rendement_moyen"] > 6) & (rend_dep["prix_m2_median"] < 5000)]
fig = px.scatter(
    rend_dep,
    x="prix_m2_median",
    y="rendement_moyen",
    size="nb_ventes",
    color="rendement_moyen",
    color_continuous_scale="Viridis",
    hover_name="Département",
    title="📈 Relation entre le prix au m² et le rendement locatif moyen par département",
    labels={
        "prix_m2_median": "Prix médian au m² (€)",
        "rendement_moyen": "Rendement brut moyen (%)"
    }
)

fig.update_layout(template="plotly_white")
fig.show()
```

```python
# Sous-ensembles typologiques
etudiant = df_rend[df_rend["nombre_pieces_principales"] == 1].copy()
jeune_couple = df_rend[df_rend["nombre_pieces_principales"].isin([2,3])].copy()

def agg_cp(df, seuil=SEUIL_VENTES):
    vr = (df.groupby(["CodePostal","Département"])
            .agg(prix_m2_median=("prix_m2","median"),
                 rendement_median=("rendement_brut","median"),
                 nb_ventes=("prix_m2","size"))
            .reset_index())
    return vr[vr["nb_ventes"] >= seuil].copy()

ville_etud = agg_cp(etudiant)
ville_couple = agg_cp(jeune_couple)

def plot_top(vr, titre):
    if vr.empty:
        print("Pas de données suffisantes pour cette typologie.")
        return
    top = vr.sort_values("rendement_median", ascending=False).head(20).reset_index(drop=True)
    fig = px.bar(
        top, x="rendement_median", y="CodePostal",
        orientation="h",
        color="prix_m2_median", color_continuous_scale="Viridis",
        text="rendement_median",
        title=titre,
        labels={"rendement_median":"Rendement médian (%)",
                "prix_m2_median":"Prix médian au m² (€)",
                "CodePostal":"Code postal"}
    )
    fig.update_traces(texttemplate="%{text:.2f}%", textposition="outside")
    fig.update_layout(template="plotly_white", height=650,
                      xaxis_title="Rendement (%)", yaxis_title="Ville / CP",
                      coloraxis_colorbar_title="€/m² (médian)")
    

```

```python
print(f"Analyse pour un profil ÉTUDIANT avec un budget max de {budget_max:,} € :")

# ========== Fusion des données ==========
dept = rend_dep.copy()

dept = pd.merge(dept, prix_dep[["Département", "Prix_m2_median", "Prix_m2_mean"]], on="Département", how="left")

# Données d’attractivité (moyenne par département)
attr_cols = ["Département", "EducationCount", "TransportCount", "CultureCount", "totmean", "AttractiviteScore"]
attr_df = dvf.groupby("Département")[attr_cols[1:]].mean().reset_index()
dept = pd.merge(dept, attr_df, on="Département", how="left")

# Rendement brut local
dept = pd.merge(dept, rendement_prix[["Département", "RendementBrut", "prix_m2"]], on="Département", how="left")

# Nettoyage
dept = dept.dropna(subset=["rendement_moyen", "TransportCount", "EducationCount", "totmean", "Prix_m2_median"])

from sklearn.preprocessing import MinMaxScaler
import numpy as np
import matplotlib.pyplot as plt

# ========== Normalisation ==========
scaler = MinMaxScaler()
cols = ["rendement_moyen", "TransportCount", "EducationCount", "CultureCount", "totmean", "Prix_m2_median", "AttractiviteScore"]
dept[[f"{c}_norm" for c in cols]] = scaler.fit_transform(dept[cols])
dept["tot_inv"] = 1 - dept["totmean_norm"]
dept["prix_inv"] = 1 - dept["Prix_m2_median_norm"]

# Rendement aplati (on ne veut pas qu'il écrase tout)
dept["rend_log"] = np.log1p(dept["rendement_moyen"] - 5)
dept["rend_norm"] = MinMaxScaler().fit_transform(dept[["rend_log"]])

# ========== Nouveau Score Étudiant ==========
# Pondérations axées sur éducation, transport, culture, prix abordable
dept["ScoreEtudiant"] = (
    0.15 * dept["rend_norm"] +           # rendement secondaire
    0.30 * dept["EducationCount_norm"] + # concentration d'établissements éducatifs
    0.25 * dept["TransportCount_norm"] + # accessibilité
    0.15 * dept["CultureCount_norm"] +   # vie étudiante / sociale
    0.10 * dept["prix_inv"] +            # abordabilité
    0.05 * dept["tot_inv"]               # cadre de vie correct
)

# ========== Pénalités ciblées ==========
# 1️⃣ Trop rural : bas prix = manque de campus ou transports
dept["penalite_rural"] = np.where(dept["Prix_m2_median"] < 1800, 0.6, 1)

# 2️⃣ Peu d'éducation = pas d'université
dept["penalite_education"] = np.where(dept["EducationCount"] < dept["EducationCount"].quantile(0.4), 0.7, 1)

# 3️⃣ Manque de transport = difficile pour les étudiants sans voiture
dept["penalite_transport"] = np.where(dept["TransportCount"] < dept["TransportCount"].quantile(0.4), 0.8, 1)

# Application des pénalités
dept["ScoreEtudiant"] *= dept["penalite_rural"] * dept["penalite_education"] * dept["penalite_transport"]

# Tri
dept = dept.sort_values("ScoreEtudiant", ascending=False).reset_index(drop=True)

# ========== Résultat principal ==========
best = dept.iloc[0]
print(f"🎓 Meilleur département pour un étudiant : {int(best['Département'])}")
print(f"  - Score global : {best['ScoreEtudiant']:.3f}")
print(f"  - Rendement moyen : {best['rendement_moyen']:.2f} %")
print(f"  - Prix médian au m² : {best['Prix_m2_median']:.0f} €/m²")
print(f"  - Education (moy) : {best['EducationCount']:.1f}")
print(f"  - Transport (moy) : {best['TransportCount']:.1f}")
print(f"  - Culture (moy) : {best['CultureCount']:.1f}")
print(f"  - Attractivité : {best['AttractiviteScore']:.2f}")
print(f"  - Cadre de vie (totmean) : {best['totmean']:.3f}")

# ========== Top 10 ==========
display_cols = [
    "Département", "ScoreEtudiant", "rendement_moyen", "Prix_m2_median",
    "EducationCount", "TransportCount", "CultureCount", "AttractiviteScore", "totmean"
]
print("\nTop 10 départements les plus adaptés aux étudiants :")

# ========== Visualisation ==========
plt.figure(figsize=(10,5))
top10 = dept.head(10).sort_values("ScoreEtudiant")
plt.barh(top10["Département"].astype(str), top10["ScoreEtudiant"], color="royalblue")
plt.xlabel("Score Étudiant (0-1)")
plt.ylabel("Département")
plt.title("Top 10 départements les plus adaptés aux étudiants (version enrichie)")
plt.tight_layout()
plt.show()
```

```python
print(f"Pour un investisseur avec un budget max de {budget_max:,} €, je recommande :")

# ========== Fusion et préparation ==========
dept_couple = rend_dep.copy()

dept_couple = pd.merge(dept_couple, prix_dep[["Département", "Prix_m2_median", "Prix_m2_mean"]], on="Département", how="left")

attr_cols = ["Département", "EducationCount", "TransportCount", "CultureCount", "totmean", "AttractiviteScore"]
attr_df = dvf.groupby("Département")[attr_cols[1:]].mean().reset_index()
dept_couple = pd.merge(dept_couple, attr_df, on="Département", how="left")

dept_couple = pd.merge(dept_couple, rendement_prix[["Département", "RendementBrut", "prix_m2"]], on="Département", how="left")

dept_couple = dept_couple.dropna(subset=["rendement_moyen", "TransportCount", "EducationCount", "totmean", "Prix_m2_median"])

from sklearn.preprocessing import MinMaxScaler
import numpy as np
import matplotlib.pyplot as plt

# ========== Normalisation ==========
scaler = MinMaxScaler()
cols = ["rendement_moyen", "TransportCount", "EducationCount", "CultureCount", "totmean", "Prix_m2_median", "AttractiviteScore"]
dept_couple[[f"{c}_norm" for c in cols]] = scaler.fit_transform(dept_couple[cols])
dept_couple["tot_inv"] = 1 - dept_couple["totmean_norm"]
dept_couple["prix_inv"] = 1 - dept_couple["Prix_m2_median_norm"]

# Rendement logarithmique pour limiter l'effet des extrêmes
dept_couple["rend_log"] = np.log1p(dept_couple["rendement_moyen"] - 4)
dept_couple["rend_norm"] = MinMaxScaler().fit_transform(dept_couple[["rend_log"]])

# ========== Nouveau Score "Jeune Couple réaliste" ==========
# 🔧 Pondérations révisées :
# - On diminue fortement le poids du rendement
# - On renforce la vie locale (transport, culture, attractivité)
# - On ajoute une pénalité spécifique aux zones sous-attractives

dept_couple["ScoreCouple"] = (
    0.15 * dept_couple["rend_norm"] +           # rendement modéré
    0.25 * dept_couple["TransportCount_norm"] + # mobilité
    0.20 * dept_couple["tot_inv"] +             # cadre de vie
    0.15 * dept_couple["prix_inv"] +            # prix abordable
    0.15 * dept_couple["CultureCount_norm"] +   # vie sociale / culture
    0.10 * dept_couple["AttractiviteScore_norm"] # dynamisme global
)

# ========== Pénalités ==========

# 1️⃣ Trop peu attractif globalement
dept_couple["penalite_attract"] = np.where(dept_couple["AttractiviteScore"] < dept_couple["AttractiviteScore"].median(), 0.8, 1)

# 2️⃣ Faible densité d'éducation/transport = isolement probable
dept_couple["penalite_isolement"] = np.where(
    (dept_couple["EducationCount"] < dept_couple["EducationCount"].quantile(0.3)) |
    (dept_couple["TransportCount"] < dept_couple["TransportCount"].quantile(0.3)),
    0.85, 1
)

dept_couple["penalite_rural"] = np.where(dept_couple["Prix_m2_median"] < 1800, 0.7, 1)

dept_couple["ScoreCouple"] *= dept_couple["penalite_attract"] * dept_couple["penalite_isolement"] * dept_couple["penalite_rural"]

dept_couple = dept_couple.sort_values("ScoreCouple", ascending=False).reset_index(drop=True)

best = dept_couple.iloc[0]
print(f"💑 Meilleur département pour un jeune couple : {int(best['Département'])}")
print(f"  - Score global : {best['ScoreCouple']:.3f}")
print(f"  - Rendement moyen : {best['rendement_moyen']:.2f} %")
print(f"  - Prix médian au m² : {best['Prix_m2_median']:.0f} €/m²")
print(f"  - Transport (moy) : {best['TransportCount']:.1f}")
print(f"  - Culture (moy) : {best['CultureCount']:.1f}")
print(f"  - Attractivité : {best['AttractiviteScore']:.2f}")
print(f"  - Cadre de vie (totmean) : {best['totmean']:.3f}")

display_cols = [
    "Département", "ScoreCouple", "rendement_moyen", "Prix_m2_median",
    "totmean", "TransportCount", "CultureCount", "AttractiviteScore"
]
print("\nTop 10 départements attractifs pour un jeune couple :")

plt.figure(figsize=(10,5))
top10 = dept_couple.head(10).sort_values("ScoreCouple")
plt.barh(top10["Département"].astype(str), top10["ScoreCouple"], color="mediumseagreen")
plt.xlabel("Score Jeune Couple (0-1)")
plt.ylabel("Département")
plt.title("Top 10 départements attractifs pour un jeune couple (scoring ajusté)")
plt.tight_layout()
plt.show()
```
